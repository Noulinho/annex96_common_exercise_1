# Thesis results: load shifting and deployment profiles

These are the saved final thesis results for 25 Vermont homes, from `ce1_occ_p2s2_all_occ`. January supplies commissioning data; February and March are deployment months. Run all cells to read the versioned CSVs and figures. This notebook does not train or simulate controllers. See [reproduction and provenance](../docs/reproduction.md) and the [artifact manifest](artifact_manifest.json).

These figures show how the final controllers redistribute heating over the day. The stage objective uses normalized temperature tracking and asymmetric guards, with weights 0.1 / 0.7 / 0.2 and 2 K scales. A time-dependent reference bias induces load shifting. Evaluation grid and comfort costs are reported separately from this training/control objective.
