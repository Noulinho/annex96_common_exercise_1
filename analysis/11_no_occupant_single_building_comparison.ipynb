# No-Occupant Single-Building CE1 Target Comparison

Compare MPC, GNU-RL / RLMPC, and SAC for the no-occupant February rollout on the notebook building:

- `resstock-amy2018-2021-release-1-247942`
- scripted result id: `building_09`

This notebook uses the **real CE1 district load target** from `data/datasets/annex96_ce1_vt_neighborhood/district_target.csv`. No carbon or price term is used because the schema has `carbon_intensity: null` and `pricing: null` for the buildings.


````python
import os
import sys
import subprocess
from pathlib import Path

os.environ.setdefault("MPLCONFIGDIR", "/tmp/matplotlib-codex")

import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", 100)
pd.set_option("display.width", 180)

ROOT = Path.cwd().resolve()
if not (ROOT / "results").exists():
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.district_target import load_ce1_district_target_profiles

BUILDING_ID = "building_09"
BUILDING_NAME = "resstock-amy2018-2021-release-1-247942"
SCENARIO = "no_occupant"
CASE = "ce1_district_target"
COMFORT_BAND = 2.0
COMFORT_SCALE = COMFORT_BAND**2
POWER_SCALE = 5.0
REWARD_WEIGHTS = {
    "comfort": 2.0,
    "energy": 0.05,
    "shift": 0.0,
    "peak": 0.0,
    "ramp": 0.05,
    "district_target": 20.0,
}
FORCE_RERUN = False

profiles = load_ce1_district_target_profiles(
    building_name=BUILDING_NAME,
    simulation_start=744,
    simulation_end=1415,
)

ROLL_OUT_PATHS = {
    "MPC": ROOT / "results/raw/mpc" / BUILDING_ID / SCENARIO / CASE / "rollout.csv",
    "GNU-RL / RLMPC": ROOT / "results/raw/rlmpc" / BUILDING_ID / SCENARIO / CASE / "rollout.csv",
    "SAC": ROOT / "results/raw/sac_merlin" / BUILDING_ID / SCENARIO / CASE / "rollout.csv",
}

SUMMARY_PATHS = {
    "MPC": ROOT / "results/summaries/mpc" / BUILDING_ID / SCENARIO / f"{CASE}_metrics.json",
    "GNU-RL / RLMPC": ROOT / "results/summaries/rlmpc" / BUILDING_ID / SCENARIO / f"{CASE}_metrics.json",
    "SAC": ROOT / "results/summaries/sac_merlin" / BUILDING_ID / SCENARIO / f"{CASE}_metrics.json",
}

KPI_PATHS = {
    "MPC": ROOT / "results/raw/mpc" / BUILDING_ID / SCENARIO / CASE / "kpis.csv",
    "GNU-RL / RLMPC": ROOT / "results/raw/rlmpc" / BUILDING_ID / SCENARIO / CASE / "kpis.csv",
    "SAC": ROOT / "results/raw/sac_merlin" / BUILDING_ID / SCENARIO / CASE / "kpis.csv",
}

print(f"Building: {BUILDING_ID} | {BUILDING_NAME}")
print(f"Case: {CASE}")
print(f"Comfort band: +/-{COMFORT_BAND:.1f} deg C around the no-occupant setpoint")
print(f"Weights: {REWARD_WEIGHTS}")
print(f"District target length: {len(profiles.target)}")
print(f"District target min/mean/max: {np.min(profiles.target):.2f} / {np.mean(profiles.target):.2f} / {np.max(profiles.target):.2f} kWh")
print(f"District background min/mean/max: {np.min(profiles.background_without_controlled_heating):.2f} / {np.mean(profiles.background_without_controlled_heating):.2f} / {np.max(profiles.background_without_controlled_heating):.2f} kWh")
print(f"Controlled heating nominal power: {profiles.heating_nominal_power:.3f} kWh/action")
print(f"District target scale: {profiles.target_scale:.3f}")

baseline_district = np.asarray(profiles.baseline_district_load, dtype=float)
baseline_error = baseline_district - np.asarray(profiles.target, dtype=float)
peak_above_baseline = np.maximum(baseline_error, 0.0).max()
print(f"Baseline district-vs-target RMSE: {np.sqrt(np.mean(baseline_error**2)):.2f} kWh")
print(f"Baseline max above-target error: {peak_above_baseline:.2f} kWh")
print(f"One-building max heating action / max above-target error: {profiles.heating_nominal_power / max(peak_above_baseline, 1e-9):.3f}")
````

## Cost / Reward

The experiment uses a shared no-price/no-carbon objective:

$$
J_t =
w_c {D_t \over \Delta T^2}
+ w_y \left({\hat P^{district}_t - P^{target}_t \over P_{scale}}\right)^2
+ w_e P^{bldg}_t
+ w_p (P^{bldg}_t)^2
+ w_r (P^{bldg}_t-P^{bldg}_{t-1})^2,
\qquad r_t=-J_t
$$

where

$$
D_t = \max(0,T_t^{ref}-\Delta T-T_t)^2 + \max(0,T_t-(T_t^{ref}+\Delta T))^2
$$

and, for this one controlled building,

$$
\hat P^{district}_t = P^{background}_t + P^{heat,controlled}_t.
$$

`P_background` is the estimated full-district baseline electricity load with building_09 baseline space-heating electricity removed. The controller then adds back its own heating electricity. This is the key fix: we compare an estimated **district** load to the CE1 **district** target, not building_09 alone to a district-level curve.


````python
COMMON_ARGS = [
    "--scenario", SCENARIO,
    "--building", BUILDING_ID,
    "--label", CASE,
    "--comfort-weight", str(REWARD_WEIGHTS["comfort"]),
    "--energy-weight", str(REWARD_WEIGHTS["energy"]),
    "--peak-weight", str(REWARD_WEIGHTS["peak"]),
    "--ramp-weight", str(REWARD_WEIGHTS["ramp"]),
    "--shift-weight", str(REWARD_WEIGHTS["shift"]),
    "--comfort-band", str(COMFORT_BAND),
    "--power-scale", str(POWER_SCALE),
    "--district-target-weight", str(REWARD_WEIGHTS["district_target"]),
    "--district-target-mode", "tracking",
]

COMMANDS = {
    "MPC": [
        sys.executable,
        "scripts/03_run_mpc.py",
        *COMMON_ARGS,
        "--objective", "balanced",
    ],
    "GNU-RL / RLMPC": [
        sys.executable,
        "scripts/05_run_rlmpc_online.py",
        *COMMON_ARGS,
        "--reward", "balanced",
    ],
    "SAC": [
        sys.executable,
        "scripts/08_run_sac_merlin_online.py",
        *COMMON_ARGS,
        "--reward", "balanced",
        "--training-episodes", "10",
    ],
}


def run_if_needed(name: str, command: list[str]):
    rollout_path = ROLL_OUT_PATHS[name]
    if rollout_path.exists() and not FORCE_RERUN:
        print(f"{name}: using existing {rollout_path.relative_to(ROOT)}")
        return
    print(f"\n=== Running {name} ===")
    print(" ".join(command))
    subprocess.run(command, cwd=ROOT, check=True)


for controller_name, command in COMMANDS.items():
    run_if_needed(controller_name, command)

for name, path in ROLL_OUT_PATHS.items():
    assert path.exists(), f"Missing rollout for {name}: {path}"
for name, path in SUMMARY_PATHS.items():
    assert path.exists(), f"Missing summary for {name}: {path}"
for name, path in KPI_PATHS.items():
    assert path.exists(), f"Missing CityLearn KPI file for {name}: {path}"
````

````python
target = np.asarray(profiles.target, dtype=float)
background = np.asarray(profiles.background_without_controlled_heating, dtype=float)
action_power = np.asarray(profiles.heating_action_power, dtype=float)
target_scale = float(profiles.target_scale)


def load_rollout(controller: str, path: Path) -> pd.DataFrame:
    df = pd.read_csv(path).copy()
    df["controller"] = controller

    has_logged_band = "tmin" in df.columns and "tmax" in df.columns
    logged_band_width = float((df["tmax"] - df["tmin"]).abs().max()) if has_logged_band else 0.0

    if "effective_setpoint" not in df.columns:
        if has_logged_band and logged_band_width > 1e-9:
            df["effective_setpoint"] = 0.5 * (df["tmin"] + df["tmax"])
        elif "tmin" in df.columns:
            df["effective_setpoint"] = df["tmin"]
        else:
            df["effective_setpoint"] = df["baseline_setpoint"]

    if "baseline_setpoint" not in df.columns:
        df["baseline_setpoint"] = df["effective_setpoint"]

    if "tmin" not in df.columns:
        df["tmin"] = df["effective_setpoint"]
    if "tmax" not in df.columns:
        df["tmax"] = df["effective_setpoint"]

    if has_logged_band and logged_band_width > 1e-9:
        df["comfort_lower"] = df["tmin"]
        df["comfort_upper"] = df["tmax"]
    else:
        df["comfort_lower"] = df["effective_setpoint"] - COMFORT_BAND
        df["comfort_upper"] = df["effective_setpoint"] + COMFORT_BAND

    if "controlled_action" not in df.columns and "action" in df.columns:
        df["controlled_action"] = df["action"]

    n = len(df)
    idx = np.minimum(np.arange(n), len(target) - 1)
    df["district_target"] = target[idx]
    df["district_background"] = background[idx]
    df["district_action_power"] = action_power[idx]

    if "heating_electricity_consumption" in df.columns:
        heating_power = np.maximum(df["heating_electricity_consumption"].to_numpy(dtype=float), 0.0)
    else:
        heating_power = np.maximum(df["controlled_action"].to_numpy(dtype=float), 0.0) * df["district_action_power"].to_numpy(dtype=float)
    df["controlled_heating_power"] = heating_power
    df["estimated_district_load"] = df["district_background"] + df["controlled_heating_power"]
    df["district_target_error"] = df["estimated_district_load"] - df["district_target"]
    df["district_above_target"] = np.maximum(df["district_target_error"], 0.0)

    df["tracking_error"] = df["indoor_temperature"] - df["effective_setpoint"]
    df["below_band"] = np.maximum(df["comfort_lower"] - df["indoor_temperature"], 0.0)
    df["above_band"] = np.maximum(df["indoor_temperature"] - df["comfort_upper"], 0.0)
    df["band_violation"] = df["below_band"] + df["above_band"]
    df["band_violation_sq"] = df["below_band"] ** 2 + df["above_band"] ** 2

    power = np.maximum(df["net_electricity_consumption"].to_numpy(dtype=float), 0.0) / POWER_SCALE
    ramp = np.diff(power, prepend=power[0])
    district_error_norm = df["district_target_error"].to_numpy(dtype=float) / target_scale
    df["normalized_grid_import"] = power
    df["candidate_stage_cost"] = (
        REWARD_WEIGHTS["comfort"] * df["band_violation_sq"] / COMFORT_SCALE
        + REWARD_WEIGHTS["district_target"] * district_error_norm**2
        + REWARD_WEIGHTS["energy"] * power
        + REWARD_WEIGHTS["peak"] * power**2
        + REWARD_WEIGHTS["ramp"] * ramp**2
    )
    return df


rollouts = {name: load_rollout(name, path) for name, path in ROLL_OUT_PATHS.items()}
combined = pd.concat(rollouts.values(), ignore_index=True)

pd.DataFrame(
    {
        "controller": list(rollouts),
        "rows": [len(df) for df in rollouts.values()],
        "columns": [", ".join(df.columns) for df in rollouts.values()],
    }
)
````

## Comfort Plots

The helper keeps the familiar `plot_comfort(...)` name, but reads from rollout CSV data. The shaded area is the soft comfort band used by the new cost/reward.


````python
def plot_comfort(history: pd.DataFrame, ax=None, title: str | None = None):
    ax = ax or plt.subplots(figsize=(12, 4))[1]
    x = history["time_step"] if "time_step" in history.columns else np.arange(len(history))
    lower = history["comfort_lower"].to_numpy(dtype=float)
    upper = history["comfort_upper"].to_numpy(dtype=float)
    tin = history["indoor_temperature"].to_numpy(dtype=float)
    tsp = history["effective_setpoint"].to_numpy(dtype=float)

    ax.plot(x, tin, label="Indoor temperature", linewidth=1.8)
    ax.plot(x, tsp, linestyle="--", label="Setpoint", linewidth=1.5, color="tab:orange")
    ax.fill_between(x, lower, upper, color="gray", alpha=0.18, label="Soft comfort band")
    ax.plot(x, lower, color="gray", alpha=0.35, linewidth=0.8)
    ax.plot(x, upper, color="gray", alpha=0.35, linewidth=0.8)

    ax.set_title(title or str(history["controller"].iloc[0]))
    ax.set_xlabel("Time step")
    ax.set_ylabel("Temperature [deg C]")
    ax.grid(True, alpha=0.3)
    ax.legend(loc="best")
    return ax


fig, axes = plt.subplots(len(rollouts), 1, figsize=(13, 9), sharex=True)
for ax, (name, df) in zip(axes, rollouts.items()):
    plot_comfort(df, ax=ax, title=f"{name} no-occupant comfort | {BUILDING_ID}")
fig.tight_layout()
plt.show()
````

## District Target Plots

These plots show the actual CE1 target against the estimated district load used in the objective:

$$\hat P^{district}_t = P^{background}_t + P^{heat,controlled}_t.$$


````python
fig, axes = plt.subplots(len(rollouts), 1, figsize=(13, 9), sharex=True)
for ax, (name, df) in zip(axes, rollouts.items()):
    x = df["time_step"] if "time_step" in df.columns else np.arange(len(df))
    ax.plot(x, df["estimated_district_load"], label="Estimated district load", linewidth=1.6)
    ax.plot(x, df["district_target"], "--", label="CE1 district target", linewidth=1.6)
    ax.fill_between(x, df["district_target"], df["estimated_district_load"], where=df["estimated_district_load"] > df["district_target"], color="tab:red", alpha=0.18, label="Above target")
    ax.set_title(f"{name} estimated district load vs target | {BUILDING_ID}")
    ax.set_ylabel("kWh")
    ax.grid(True, alpha=0.3)
    ax.legend(loc="best")
axes[-1].set_xlabel("Time step")
fig.tight_layout()
plt.show()
````

## KPI Tables

The first table uses the JSON summaries from the scripts. The second table recomputes target-tracking, comfort, and load KPIs from the rollout CSVs.


````python
SUMMARY_COLUMNS = [
    "reward_total",
    "electricity_consumption_total",
    "daily_peak_average",
    "discomfort_proportion",
    "discomfort_cold_proportion",
    "discomfort_hot_proportion",
    "district_target_weight",
    "district_target_scale",
    "controlled_heating_nominal_power",
    "override_count",
]

summary_rows = []
for controller, path in SUMMARY_PATHS.items():
    data = json.loads(path.read_text())
    row = {"controller": controller}
    row.update({col: data.get(col, np.nan) for col in SUMMARY_COLUMNS})
    row["building"] = data.get("building")
    row["building_name"] = data.get("building_name")
    summary_rows.append(row)

summary_kpis = pd.DataFrame(summary_rows).set_index("controller")
summary_kpis
````

````python
def rollout_kpis(df: pd.DataFrame) -> dict[str, float]:
    p = df["net_electricity_consumption"].to_numpy(dtype=float)
    action = df["controlled_action"].to_numpy(dtype=float)
    err = df["tracking_error"].to_numpy(dtype=float)
    violation = df["band_violation"].to_numpy(dtype=float)
    district_error = df["district_target_error"].to_numpy(dtype=float)
    above = df["district_above_target"].to_numpy(dtype=float)
    stage_cost = df["candidate_stage_cost"].to_numpy(dtype=float)
    return {
        "building_grid_energy_sum": float(np.nansum(p)),
        "building_grid_peak": float(np.nanmax(p)),
        "building_ramping_abs_sum": float(np.nansum(np.abs(np.diff(p)))),
        "controlled_heating_sum": float(np.nansum(df["controlled_heating_power"])),
        "estimated_district_energy_sum": float(np.nansum(df["estimated_district_load"])),
        "estimated_district_peak": float(np.nanmax(df["estimated_district_load"])),
        "target_rmse_kWh": float(np.sqrt(np.nanmean(district_error**2))),
        "target_mae_kWh": float(np.nanmean(np.abs(district_error))),
        "above_target_kWh_sum": float(np.nansum(above)),
        "above_target_peak_kWh": float(np.nanmax(above)),
        "above_target_fraction": float(np.nanmean(above > 1e-9)),
        "candidate_cost_sum": float(np.nansum(stage_cost)),
        "recomputed_reward_total": float(-np.nansum(stage_cost)),
        "action_mean": float(np.nanmean(action)),
        "action_sum": float(np.nansum(action)),
        "tracking_mae_degC": float(np.nanmean(np.abs(err))),
        "tracking_rmse_degC": float(np.sqrt(np.nanmean(err**2))),
        "comfort_violation_fraction": float(np.nanmean(violation > 1e-9)),
        "comfort_violation_degC_sum": float(np.nansum(violation)),
        "comfort_violation_sq_sum": float(np.nansum(df["band_violation_sq"])),
    }


direct_kpis = pd.DataFrame.from_dict(
    {name: rollout_kpis(df) for name, df in rollouts.items()},
    orient="index",
)
direct_kpis
````

````python
selected_citylearn_metrics = [
    "electricity_consumption_total",
    "daily_peak_average",
    "ramping_average",
    "discomfort_proportion",
    "discomfort_cold_proportion",
    "discomfort_hot_proportion",
]

citylearn_tables = []
for controller, path in KPI_PATHS.items():
    kpis = pd.read_csv(path)
    rows = kpis[(kpis["level"] == "district") & (kpis["cost_function"].isin(selected_citylearn_metrics))]
    citylearn_tables.append(rows.assign(controller=controller)[["controller", "cost_function", "value"]])

citylearn_kpis = pd.concat(citylearn_tables, ignore_index=True)
citylearn_pivot = citylearn_kpis.pivot(index="controller", columns="cost_function", values="value")
citylearn_pivot[selected_citylearn_metrics]
````

## Notes

- This notebook uses the `ce1_district_target` results, not the older `peak_aware_shift` pseudo-stress experiment.
- The load-shifting / peak-shaving signal is the real CE1 `district_load_target` trajectory.
- The objective has no price or carbon term because those signals are not active in the VT CE1 schema.
- GNU-RL / RLMPC uses the same target-aware inner stage-cost structure as MPC for this no-occupant run. Because this matched-cost mode fixes the MPC cost weights, the PPO cost-parameter update is skipped for this specific comparison.



## CE1 Exact-Setpoint Normalized Shift Run

Appended run label: `ce1_setpoint_norm_shift20`.

Cost used by MPC and matched GNU-RL/RLMPC:

\[
J_k =
30\left(\frac{T_k - T^{{sp}}_k}{\sigma_T}\right)^2
+ 0.05u_k
+ 2u_k^2
+ 0.1(u_k-u_{k-1})^2
+ 20\lambda_k u_k
+ 10\left(\frac{L^{{base-bg}}_k + P^{{heat}}_k - L^{{CE1}}_k}{\sigma_L}\right)^2 .
\]

Normalization:

- `Tsp` is the single dataset heating setpoint; `tmin = tmax = Tsp` in the saved rollouts.
- `sigma_T` is the building-specific January scheduled-setpoint standard deviation (`setpoint_std`), not a hand divisor.
- `u_k` is already min-max normalized by the actuator bounds `[0, 1]`.
- `lambda_k = max(L_baseline - L_CE1, 0) / max(max(L_baseline - L_CE1, 0))`.
- `sigma_L` is the February baseline district-load target-error standard deviation.

Weights:

```json
{
  "label": "ce1_setpoint_norm_shift20",
  "comfort_mode": "setpoint",
  "temperature_scale_mode": "setpoint_std",
  "district_target_scale_mode": "baseline_error_std",
  "comfort_weight": 30.0,
  "energy_weight": 0.05,
  "peak_weight": 2.0,
  "ramp_weight": 0.1,
  "shift_weight": 20.0,
  "district_target_weight": 10.0
}
```

## CE1 Obs-Minmax Peak/Shift Run

Appended run label: `ce1_obsminmax_c3_peak25_shift30_target30`.

Cost used by MPC and matched GNU-RL/RLMPC:

\[
J_k =
3\left(\frac{T_k - T^{sp}_k}{T^{max}_{obs}-T^{min}_{obs}}\right)^2
+ 25u_k^2
+ (u_k-u_{k-1})^2
+ 30\lambda_k u_k
+ 30\left(\frac{L^{bg}_k + P^{heat,max}_k u_k - L^{CE1}_k}{L^{err,max}_{ctrl}}\right)^2 .
\]

Normalization:

- `Tsp` is the single CE1 heating setpoint; saved rollouts use `tmin = tmax = Tsp`.
- Temperature uses CityLearn observation-space min/max bounds, not a standard deviation and not a hand divisor.
- `u_k`, `u_k^2`, `(u_k-u_{k-1})^2`, and `lambda_k u_k` are normalized by construction because `u_k in [0,1]` and `lambda_k in [0,1]`.
- `L_err,max_ctrl` is the min/max controllable district target-error bound over `u in [0,1]`.
- The energy linear term is off in this run (`w_energy = 0`) so peak and shift dominate the load objective.

Weights:

```json
{
  "label": "ce1_obsminmax_c3_peak25_shift30_target30",
  "comfort_mode": "setpoint",
  "temperature_scale_mode": "observation_minmax",
  "district_target_scale_mode": "controllable_error_minmax",
  "comfort_weight": 3.0,
  "energy_weight": 0.0,
  "peak_weight": 25.0,
  "ramp_weight": 1.0,
  "shift_weight": 30.0,
  "district_target_weight": 30.0
}
```

## CE1 Asymmetric Comfort + Peak/Shift Run

Appended run label: `ce1_preheat_credit_v1`.

MPC and matched GNU-RL/RLMPC minimize this weighted stage cost:

\[
J_k = J^T_k
+ w_p p_k^2
+ w_r(p_k-p_{k-1})^2
+ w_s\lambda_k p_k
+ w_L\left[\max\left(0,\frac{L^{bg}_k+P^{ctrl}_k-L^{CE1}_k}{S_L}\right)\right]^2
- w_{pre}(1-\lambda_k)\lambda^{future}_k p_k .
\]

For MPC/GNU-RL, \(p_k=u_k\) and \(P^{ctrl}_k=P^{heat,max}_k u_k\).  
For SAC, reward is \(r_k=-J_k\), \(p_k=P^{heat}_k/P^{heat,max}\), and \(P^{ctrl}_k=P^{heat}_k\).

Comfort uses the true CityLearn setpoint, not `setpoint - 2`:

\[
J^T_k =
w_{track}\left(\frac{T_k-T^{sp}_k}{s_{track}}\right)^2
+w_{cold}\left(\frac{\max(0,T^{sp}_k-\Delta_{cold}-T_k)}{s_{cold}}\right)^2
+w_{hot}\left(\frac{\max(0,T_k-T^{sp}_k-\Delta_{hot})}{s_{hot}}\right)^2 .
\]

Here \(\lambda^{future}_k\) is the maximum CE1 stress in the look-ahead preheat window.
So the controller is explicitly credited for heating during low-stress hours before future stress, while cold and hot comfort violations remain penalized.
The district target term is `peak_only`: below-target load is not penalized by the target term; only above-target load is.

## CE1 Asymmetric Hot-Heavy + Shift/Preheat v1 — MPC, GNU-RL/RLMPC, SAC

Appended run label: `ce1_asym_hot_heavy_shift_v1`.

Same asymmetric hot-heavy comfort as `ce1_asym_hot_heavy_v1`, plus explicit shift and preheat signals:
- **Shift penalty** (\(w_s=0.3\)): penalizes heating proportionally to grid stress \(\lambda_k\) — discourages heating when the grid is already stressed
- **Preheat credit** (\(w_{pre}=0.3\)): rewards heating during low-stress hours before future high-stress — encourages midnight preheating

Stage cost:
\[
J_k = J^T_k + w_{du}(u_k-u_{k-1})^2 + w_s\lambda_k p_k - w_{pre}(1-\lambda_k)\lambda^{future}_k p_k + w_L\left(\frac{L^{bg}_k+P^{ctrl}_k-L^{CE1}_k}{S_L}\right)^2
\]

## CE1 Time-Biased Setpoint Guard 2R2C+Kalman - `ce1_time_bias_guard_1c_2r2c_kalman`

Run labels: `ce1_time_bias_guard_1c_2r2c_kalman` for MPC and GNU-RL/RLMPC with January-only 2R2C RC models and a Kalman latent-mass observer; `ce1_time_bias_guard_1c_2r2c_kalman_bc_mpc_only_es_a1` for PI + behavior cloning from the 2R2C+Kalman MPC teacher; `ce1_time_bias_guard_1c_2r2c_kalman_bc_mpc_sac10_es_a1` for the same BC initialization followed by 10 SAC reward-training episodes. Both SAC runs use `action_scaling_coefficient=1.0`.

Cost/reward used for the new GNU-RL/RLMPC and SAC runs:

`J_k = 0.5 * [3*((Tin-(Tsp+bias))/2)^2 + 60*(cold_guard/1.5)^2 + 15*(hot_guard/2)^2] + 0.5*GridStress*P_heat/5`

`reward = -J_k`, `bias=+1 C` in preheat windows, `bias=-1 C` in reduce windows, and `cold_guard/hot_guard` are measured against the true setpoint with a `Tsp +/- 2 C` guard. Comfort and time-weighted heating weights sum to 1.0.

GNU-RL/RLMPC uses the 2R2C DiffMPC checkpoint trained on January PI data. Because qpth emitted `24148` inaccurate-solution warnings across offline training and February deployment, this row should be treated as informative but numerically less clean than explicit MPC/SAC.

The BC rows keep the January PI replay episode, train the SAC actor offline against January 2R2C+Kalman MPC action labels with validation early stopping (`patience=5`), then either deploy immediately or continue with 10 SAC reward-training episodes before February deployment.

## CE1 Safe-PI Reset Penalty MERLIN SAC - `ce1_time_bias_guard_1c_1r1c_sac_safepi_penalty100_auto_a1`

Run label: `ce1_time_bias_guard_1c_1r1c_sac_safepi_penalty100_auto_a1`. SAC-only ablation, no MPC teacher and no behavior cloning.

Pipeline: January episode 0 uses the safe PI replay-fill controller; the following January episodes use SAC reward training. During January training, runaway LSTM temperatures are still reset back near the setpoint. Safe-PI/bootstrap reset transitions are dropped as untrusted LSTM cleanup data, but **SAC-caused reset transitions are kept in replay with an additive `-100` reward penalty**. February deployment is not reset or shielded, so February KPIs are honest deployment results.

SAC settings: `action_scaling_coefficient=1.0`, automatic entropy tuning, robust replay normalization with observation/reward clipping, structured exploration, and cold comfort shield during training. Reward/cost is unchanged:

`J_k = 0.5 * [3*((Tin-(Tsp+bias))/2)^2 + 60*(cold_guard/1.5)^2 + 15*(hot_guard/2)^2] + 0.5*GridStress*P_heat/5`

`reward = -J_k`, `bias=+1 C` in preheat windows, `bias=-1 C` in reduce windows, and guards use true `Tsp +/- 2 C`.

Cost/reward:

`D_k = 3*((Tin-(Tsp+bias))/2)^2 + 60*(max(Tsp-2-Tin,0)/1.5)^2 + 15*(max(Tin-Tsp-2,0)/2)^2`

Optimizer-side MPC/RLMPC proxy cost:

`J_proxy,k = 0.5*D_k + 0.5*GridStress*(p_nom*u_k)/5`

Environment-side reward and diagnostics:

`J_real,k = 0.5*D_k + 0.5*GridStress*P_heat_env/5`, `reward = -J_real,k`.

`bias=+1 C` in preheat windows, `bias=-1 C` in reduce windows, and `GridStress = baseline_district_daily_mean_exp2_norm`. The common term diagnostics below use the realized environment heat for all controllers. This keeps SAC/RLMPC reward accounting honest while still acknowledging that explicit MPC optimizes the action-power proxy inside the horizon.

## Final No-Occupant Normalized-D MPC/RLMPC/SAC - `ce1_final_no_occ_normD010702_controller_comparison`

Final no-occupant comfort-only comparison after dropping the GridStress/load term.

Controller labels:

`ce1_final_no_occ_D010702_mpc_2r2c_kalman`: MPC with January-only 2R2C+Kalman RC fit, horizon 24, deployed in February.

`ce1_final_no_occ_D010702_rlmpc_1r1c_h6`: GNU-RL/RLMPC using the 1R1C horizon-6 pipeline that previously gave the stable RLMPC behavior, with the same normalized comfort reward.

`ce1_final_no_occ_D010702_sac_safepi9_a1` plus `ce1_final_no_occ_D010702_sac_safepi9_bad10_a1`: SAC safe-PI bootstrap on January, then 9 reward-training episodes. Buildings with February discomfort above 10% after episode 9 were resumed for 10 more reward-training episodes. The final SAC row is the hybrid final deployment policy. This SAC run was launched outside the sandbox so PyTorch could use CUDA.

Cost/reward used here:

`D_k = 0.1*((Tin-(Tsp+bias))/2)^2 + 0.7*(max(Tsp-2-Tin,0)/1.5)^2 + 0.2*(max(Tin-Tsp-2,0)/2)^2`

`J_k = D_k`, `reward = -D_k`.

There is no GridStress term, no electricity term, no peak term, and no ramp term in this final no-occupant objective. The three displayed weights sum to 1.0. `bias=+1 C` in the preheat windows, `bias=-1 C` in the reduce windows, and cold/hot guard penalties are measured against the true setpoint with a `Tsp +/- 2 C` guard.

## Final No-Occupant Normalized-D MPC/SAC/Fixed-Adaptive GNU-RL - `ce1_final_no_occ_normD010702_fixed_adaptive_rlmpc_comparison`

Final no-occupant comfort-only comparison for February and March.

Controller labels:

`ce1_final_no_occ_D010702_mpc_2r2c_kalman` / `ce1_final_no_occ_D010702_mpc_2r2c_kalman_mar`: MPC with January-only 2R2C+Kalman RC fit, horizon 24, deployed in February and March.

`ce1_final_no_occ_D010702_rlmpc_1r1c_h6_fixed` / `ce1_final_no_occ_D010702_rlmpc_1r1c_h6_fixed_mar`: fixed GNU-RL/RLMPC with the completed 1R1C imitation-learned model, horizon 6, and fixed comfort weights `(0.1, 0.7, 0.2)`.

`ce1_final_no_occ_D010702_rlmpc_1r1c_h6_adaptive_alpha` / `ce1_final_no_occ_D010702_rlmpc_1r1c_h6_adaptive_alpha_mar`: same GNU-RL/RLMPC pipeline, but PPO updates only the inner-MPC softmax comfort weights `(alpha_track, alpha_cold, alpha_hot)` online during the deployment month.

`ce1_final_no_occ_D010702_sac_safepi9_a1` + `ce1_final_no_occ_D010702_sac_safepi9_bad10_a1`: SAC safe-PI bootstrap on January, 9 SAC reward episodes, then 10 extra episodes for bad-comfort buildings. `ce1_final_no_occ_D010702_sac_feb10_mar_a1` continues SAC training for 10 February episodes and deploys in March.

Cost/reward used here:

`D_k = 0.1*((Tin-(Tsp+bias))/2)^2 + 0.7*(max(Tsp-2-Tin,0)/1.5)^2 + 0.2*(max(Tin-Tsp-2,0)/2)^2`

`J_k = D_k`, `reward = -D_k`.

There is no GridStress term, no electricity term, no peak term, and no ramp term. The weights sum to 1.0. `bias=+1 C` in preheat windows, `bias=-1 C` in reduce windows, and cold/hot guards use the true setpoint with a `Tsp +/- 2 C` band.

## Final No-Occupant Normalized-D MPC/SAC/GNU-RL Dynamics Adaptation - `ce1_final_no_occ_normD010702_dynamics_rlmpc_comparison`

Final no-occupant comfort-only comparison with the fixed and adaptive GNU-RL/RLMPC variants.

Controller labels:

`ce1_final_no_occ_D010702_mpc_2r2c_kalman`: MPC with January-only 2R2C+Kalman RC fit, horizon 24, deployed in February.

`ce1_final_no_occ_D010702_rlmpc_1r1c_h6_fixed`: fixed GNU-RL/RLMPC using the completed 1R1C imitation-learned RC policy, horizon 6, and the same normalized comfort-only inner MPC cost. PPO is collected but the inner cost weights stay fixed at `(0.1, 0.7, 0.2)`, so the online cost update is skipped.

`ce1_final_no_occ_D010702_rlmpc_1r1c_h6_adaptive_dyn`: adaptive-dynamics GNU-RL/RLMPC with the same 1R1C horizon-6 pipeline and same fixed reward/cost, but after the stochastic February rollout it updates the differentiable RC dynamics from one-step `Tin(k+1)` prediction error. The comfort objective is not changed.

`ce1_final_no_occ_D010702_sac_safepi9_a1` plus `ce1_final_no_occ_D010702_sac_safepi9_bad10_a1`: SAC safe-PI bootstrap on January, then 9 reward-training episodes. Buildings with February discomfort above 10% after episode 9 were resumed for 10 more reward-training episodes. The final SAC row is the hybrid final deployment policy.

Cost/reward used here:

`D_k = 0.1*((Tin-(Tsp+bias))/2)^2 + 0.7*(max(Tsp-2-Tin,0)/1.5)^2 + 0.2*(max(Tin-Tsp-2,0)/2)^2`

`J_k = D_k`, `reward = -D_k`.

There is no GridStress term, no electricity term, no peak term, and no ramp term. The three comfort weights sum to 1.0. `bias=+1 C` in the preheat windows, `bias=-1 C` in the reduce windows, and cold/hot guard penalties are measured against the true setpoint with a `Tsp +/- 2 C` guard.

Final occupant-in-loop run with cyclic occupant assignment:
B1/B5/... use `occ1_tolerant`, B2/B6/... use `occ2_sensitive`, B3/B7/... use `occ3_cold`, and B4/B8/... use `occ4_hot`.

Common cost/reward for all controllers:

`D_k = 0.1*((Tin-(Tref+bias))/2)^2 + 0.7*(max(Tref-2-Tin,0)/1.5)^2 + 0.2*(max(Tin-Tref-2,0)/2)^2`

`reward = -D_k`, no GridStress/energy term. `bias=+1 C` in preheat windows, `bias=-1 C` in reduce windows, and the true-reference guard is `Tref +/- 2 C`.

Three deployment references are shown where the artifacts exist: no occupant (`no_occupant`), raw occupant overrides (`without_tdyn`), and B-spline occupancy-predicted adaptive reference tracking (`bspline_tdyn`). GNU-RL has two fixed-cost variants: fixed model, and online RC-model adaptation from one-step `Tin(k+1)` prediction error. SAC uses one safe-PI January episode plus 19 January reward-training episodes for February deployment, then resumes for 10 February reward-training episodes for March deployment.

The B-spline Tdyn branch uses the Zhang-style dynamic reference update:

`Tdyn_{k+1} = clip(Tdyn_k + rho*(Tpref-Tdyn_k) + bounded_feedback_nudge_toward_Tpref)`

and the controller/reward reference is occupancy-gated:

`Tref_k = (1 - p_occ,k)*T_schedule,k + p_occ,k*Tdyn_k`.

## CE1 Thesis-Clean Load Figures - `ce1_thesis_clean_load_figures`

Thesis-clean load-shape figures.

Figure A keeps the no-occupant comparison readable: CityLearn baseline, the new
Safe-PI/RBC reference baseline, MPC, GNU-RL, and SAC. Missing curves are marked
on the plot instead of silently mixing old-cost artifacts.

The February/March Figure A panel stacks the two deployment months vertically so
the load-shifting behavior can be compared directly without the January
jump-start period.

The January SAC curve is now a deterministic deployment of the exact actor
trained on January (`1` Safe-PI/MERLIN replay episode + `19` SAC reward-training
episodes), evaluated back on January with no exploration and no additional
training.

The Safe-PI/RBC baseline is **not** CityLearn's `BaselineAgent`. It is a separate
simple controller that tracks:

`T_control_ref = Tref + bias`

with:

`Tref = Tschedule` for no occupant,
`Tref = effective occupant setpoint` for raw occupant overrides, and
`Tref = (1-occ_bin)*Tschedule + occ_bin*Tdyn` for B-spline Tdyn.

The time-bias windows are `+1 C` in hours 1-5 and 14-17, `-1 C` in hours 7-10
and 18-22, using CE1 1..24 hour labels.

Figure B keeps the occupant question separate: one controller per plot, comparing
`no_occupant` against `occ1` through `occ4` under the B-spline Tdyn branch.

The Safe-PI/RBC comfort grids below use the same compact `plot_comfort` style:
blue is `Tin`, orange is the tracked setpoint (`Tref+bias`), gray is the
`Tref +/- 2 C` guard band, and the B-spline Tdyn cases also show `Tdyn` as a
solid black line.

The cumulative deployment-type KPI trajectory is computed at every rollout
timestep from the district load traces. January contains only CityLearn baseline
and Safe-PI. MPC, GNU-RL, and SAC are not plotted in January; their cumulative
trajectories start at the February boundary after sharing the January Safe-PI
history.

The non-cumulative daily KPI trajectory computes each day's KPI from that day's
24-hour load profile and normalizes it against CityLearn baseline for the same
day. The dashed black line is therefore exactly the CityLearn baseline score of
`1.0`.

## CE1 Supplementary KPI Cost Audit - `ce1_occ_p2s2_all_occ`

Supplementary KPI audit using the definitions from *Supplementary materials for "Optimizing building energy systems for grid-interactivity, comfort and resilience"* and the formulas implemented in `citylearn/cost_function.py` and `CityLearnEnv.evaluate()`.

Important distinction: the controller stage reward is still the normalized temperature-tracking cost used in the control experiments. The new `S_c` and `S_g` below are **evaluation costs**, not a new controller reward.

Formula/code match:

- `S_c` comfort cost maps to CityLearn `discomfort_proportion`, averaged over the 25 buildings. With occupant counts, CityLearn evaluates discomfort over occupied/evaluated steps.
- `S_g = 0.25*(S_ramp + S_1-LF + S_daypk + S_peak)`.
- `S_ramp` maps to CityLearn `ramping_average`. In this CityLearn code path, `CostFunction.ramping(..., down_ramp=None)` defaults to `down_ramp=False`, so it is cumulative positive **up-ramp** normalized by baseline.
- `S_1-LF` maps to `daily_one_minus_load_factor_average`.
- `S_daypk` maps to `daily_peak_average`.
- `S_peak` maps to `all_time_peak_average`.
- The raw `ramping_abs_kwh` column remains a separate auxiliary metric and is not the CityLearn `S_ramp` score.

## CE1 Thesis Final Compact Results - `ce1_thesis_final_results`

This section condenses the final result artifacts into thesis-ready tables and
figures. It does not rerun any controller.

Representative buildings are fixed once across the full result set:

- **Building 12**: good no-occupant comfort and stable RC fit.
- **Building 15**: middle no-occupant comfort tail.
- **Building 8**: poor RC fit and poor no-occupant comfort for MPC and SAC.

The no-occupant comparison includes CityLearn baseline, Safe-PI/RBC, MPC,
adaptive GNU-RL, and SAC. The occupant-analysis figures then focus on MPC after
the controller comparison, because MPC is the most consistent comfort controller
under B-spline $T^{dyn}$. The occupant plots reuse the same three buildings so the
occupant effect is not confounded by changing examples.

Final occupant-in-loop run with **each occupant evaluated on all 25 buildings**. This replaces the earlier cyclic occupant assignment.

Cases:

- `no_occupant`: no occupant model, no replay-tail for SAC.
- `without_tdyn`: occupant overrides in the loop, controller tracks the active/effective setpoint.
- `bspline_tdyn`: occupant overrides plus B-spline/Tdyn reference tracking. SAC uses the gentle replay-tail for February-to-March continuation only in this Tdyn branch.

Common cost/reward for all controllers:

`D_k = 0.1*((Tin-(Tref+bias))/2)^2 + 0.7*(max(Tref-2-Tin,0)/2)^2 + 0.2*(max(Tin-Tref-2,0)/2)^2`

`reward = -D_k`, no GridStress/energy term. `bias=+1 C` in preheat windows, `bias=-1 C` in reduce windows, and the true-reference guard is `Tref +/- 2 C`.

MPC uses the existing January-fitted 2R2C+Kalman RC models. GNU-RL uses the existing January imitation checkpoints; the adaptive variant starts March from the February online-updated model and continues model-parameter adaptation. SAC uses one safe-PI January episode plus 19 January reward-training episodes for February deployment, then resumes for 10 February reward-training episodes for March deployment.

The B-spline branch uses the Zhang-style dynamic state update:

`Tdyn_{k+1} = clip(Tdyn_k + rho*(Tpref-Tdyn_k) + bounded_feedback_nudge_toward_Tpref)`

and the controller/reward reference is occupancy-gated:

`Tref_k = (1 - occ_bin,k)*T_schedule,k + occ_bin,k*Tdyn_k`.

So predicted unoccupied means the controller returns to the schedule; predicted occupied means it tracks Tdyn. In the comfort plots, the orange line is this gated controller reference and the black line is the underlying Tdyn state.

KPI naming:

- Building-level distributions use CityLearn scores where available: imported-electricity score, zero-net-energy score, average-daily-peak score, maximum-peak-demand score, up-ramping score, daily/monthly one-minus load-factor scores, and CityLearn discomfort proportions.
- District-level trajectories use the aggregate district net-electricity demand `P_k=sum_b P_el,b,k`; imported electricity is the positive/imported part `sum max(P_k,0)`.
- CityLearn `ramping_average` is cumulative **up-ramping** because `CostFunction.ramping(..., down_ramp=None)` defaults to `down_ramp=False`. The auxiliary `ramping_abs_kwh` column is the Zoltan-style absolute smoothness metric `sum |P_k-P_{k-1}|`.
- `daily_one_minus_load_factor_average` is the daily flatness score (`window=24`); `monthly_one_minus_load_factor_average` is reported only as an auxiliary building-level score.
- Added non-CityLearn auxiliary columns are raw building net electricity, input heating demand, HVAC heating electricity, Kelvin-hours, and override-event counts.
- In this dataset/artifact set, `cost_total` and `carbon_emissions_total` may be blank when CityLearn pricing or carbon-intensity inputs are unavailable/zero.

ISSO 74 ATG comfort-class summary:

The ATG bands use `Tneutral = 21.45 + 0.11*Teref`, with `Teref = (Tout_today + 0.8*Tout_yesterday + 0.4*Tout_2_days_ago + 0.2*Tout_3_days_ago)/2.4`. A building is labelled A/B/C here by time-compliance: A if at least 90% of hours are inside `Tneutral +/- 1.25 C`, B if at least 80% are inside `+/- 2.0 C`, C if at least 65% are inside `+/- 2.5 C`; otherwise `outside_C`. The `%` columns report the mean share of hours inside each band across the 25 buildings.
